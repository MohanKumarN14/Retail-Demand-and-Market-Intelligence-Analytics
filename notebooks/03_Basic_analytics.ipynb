```python
import shutil
from pathlib import Path

# Original cleaned dataset — DO NOT MODIFY
source = Path("../data/processed/online_retail_II_cleaned.parquet")

# Separate copies for this basic analytics project
parquet_destination = Path(
    "../data/basic_project/online_retail_II_basic.parquet"
)

csv_destination = Path(
    "../data/basic_project/online_retail_II_basic.csv"
)

# Create basic_project folder if it doesn't exist
parquet_destination.parent.mkdir(parents=True, exist_ok=True)

# 1. Copy the original Parquet file
shutil.copy2(source, parquet_destination)

# 2. Read the copied Parquet file
import pandas as pd

df = pd.read_parquet(parquet_destination)

# 3. Create a CSV copy
df.to_csv(csv_destination, index=False)

print("Copies created successfully:")
print(parquet_destination)
print(csv_destination)
```
