# Display StockCodes that contain alphabetic characters


```python
retail[
    retail["StockCode"].str.contains("[A-Za-z]", regex=True, na=False)
]["StockCode"].value_counts()
```

```python
retail[retail["StockCode"].str.fullmatch(r"[A-Za-z]+", na=False)]["StockCode"].value_counts()
```

**Observation:** The investigation identified several StockCodes consisting entirely of alphabetic characters that represent administrative or operational transactions rather than physical retail products. Examples include POST (Postage), M (Manual Adjustment), D (Discount), B (Bad Debt Adjustment), and AMAZONFEE. These entries should be treated separately from product sales during the Data Cleaning phase. A few alphabetic StockCodes (e.g., DCGSSGIRL, DCGSSBOY, PADS, and GIFT) may still represent genuine products or promotional items and should be verified before any exclusions are made.

```python
retail[retail["Quantity"].isin([retail["Quantity"].max(), retail["Quantity"].min()])]
```

**Observation:** The largest positive and negative Quantity values (±80,995) belong to the same product, customer, and price, with the negative transaction recorded as a cancelled invoice shortly after the original purchase. This indicates that the extreme values represent a legitimate business transaction and its corresponding cancellation rather than a data entry error. The handling of cancelled transactions will be addressed during the Data Cleaning phase.

```python
retail[retail["Price"] < 0]
```

**Observation:** All rows with negative prices correspond to the administrative StockCode 'B' with the description 'Adjust bad debt'. These are accounting adjustments rather than actual product sales. In addition, all such records have missing Customer IDs, indicating that they cannot be linked to individual customers. These entries will require special handling during the Data Cleaning phase and should not be treated as normal retail transactions.

```python
retail[retail["StockCode"].str.fullmatch(r"[A-Za-z]+", na=False)]["StockCode"].value_counts()
```

**Observation:** The investigation identified several StockCodes consisting entirely of alphabetic characters that represent administrative or operational transactions rather than physical retail products. Examples include POST (Postage), M (Manual Adjustment), D (Discount), B (Bad Debt Adjustment), and AMAZONFEE.
```python
(retail.groupby("StockCode")["Description"].nunique().sort_values(ascending=False).head(20))
```

**Observation:** Some StockCodes are associated with multiple Description values due to data entry mistakes and manual correction records rather than representing different products.

```python
retail[retail.duplicated(keep=False)].sort_values(["Invoice", "StockCode"]).head(20)
```

**Observation:** Inspection of duplicate records confirms that all duplicated rows are identical across every column, including the invoice number, product, quantity, price, timestamp, customer, and country. These duplicates are therefore likely caused by data entry or system duplication rather than genuine repeated purchases. Retaining them would result in double-counting of sales, revenue, and product demand. These records should be removed during the Data Cleaning phase.
